```python
data_123 = pd.concat([data_12, data3])
data_123 = data_123.drop_duplicates()
data_123.head()
```

```python
data_123.info()
```

```python
data3['Agent_name'] = data3['Agent_name'].apply(
    lambda x: x.split()[0][0] + ". " + x.split()[1] if isinstance(x, str) else np.NAN
)
```

```python
data_123 = pd.concat([data_12, data3])
```

```python
orders_data = pd.read_csv('./data/orders.csv', index_col=0)
orders_data.head()
```

```python
orders_data = orders_data.rename(columns={
    'Id': 'Order_id',
})
```

```python
data_with_orders = data_123.merge(orders_data, on='Order_id', how='left')
data_with_orders.head()
```

```python

total_orders_in_requests = data_123['Order_id'].notna().sum()
matching_orders = data_123['Order_id'].isin(orders_data['Order_id']).sum()

print(f"non null order irds in support data: {total_orders_in_requests}")
print(f"number of these order ids actually present in orders_data: {matching_orders}")
print(f"percentage: {(matching_orders / total_orders_in_requests * 100):.2f}%\n")


missing_example_id = data_123[~data_123['Order_id'].isin(orders_data['Order_id'])]['Order_id'].dropna().iloc[0]
print(f"point check for order id: {missing_example_id}")
print("Is it in data_123?", (data_123['Order_id'] == missing_example_id).any())
print("Is it in orders_data?", (orders_data['Order_id'] == missing_example_id).any())
```
